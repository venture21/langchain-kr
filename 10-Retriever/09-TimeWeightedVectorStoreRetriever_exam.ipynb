```python
import os
from datetime import datetime, timedelta
from langchain.retrievers import TimeWeightedVectorStoreRetriever
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
from langchain_core.documents import Document
from dotenv import load_dotenv

# API 키 정보 로드
load_dotenv()
```

```python
# 예제 문서와 시간 정보
docs = [
    Document(
        page_content="2월: 정부의 의대 정원 증원 결정에 반발한 전공의들이 집단 사직하며 의료 공백 사태가 발생했다.",
        metadata={"created_at": "2024-02-01"},
    ),
    Document(
        page_content="4월 10일: 제22대 국회의원 총선거에서 야당인 더불어민주당이 압승하며 여소야대 정국이 이어졌다.",
        metadata={"created_at": "2024-04-10"},
    ),
    Document(
        page_content="5월: K-드라마와 K-POP이 국제 OTT 플랫폼에서 큰 인기를 끌며 한류 콘텐츠의 영향력이 더욱 확대되었다.",
        metadata={"created_at": "2024-05-01"},
    ),
    Document(
        page_content="6월: 반도체 산업이 AI 수요에 힘입어 역대급 수출 실적을 기록하며 한국 경제 성장을 견인했다.",
        metadata={"created_at": "2024-06-01"},
    ),
    Document(
        page_content="7월: 2024 파리 올림픽에서 대한민국 선수단이 좋은 성적을 거두며 국민들에게 감동을 선사했다.",
        metadata={"created_at": "2024-07-01"},
    ),
    Document(
        page_content="8월: 북한이 군사정찰위성 발사를 강행하고 오물 풍선을 살포하는 등 남북 간 긴장감이 고조되었다.",
        metadata={"created_at": "2024-08-01"},
    ),
    Document(
        page_content="9월: 고물가와 고금리 상황이 지속되면서 서민 경제의 부담이 가중되었다.",
        metadata={"created_at": "2024-09-01"},
    ),
    Document(
        page_content="10월: 정부가 인공지능(AI)과 디지털 산업 육성을 위한 새로운 정책 패키지를 발표했다.",
        metadata={"created_at": "2024-10-01"},
    ),
    Document(
        page_content="11월: 한국형 발사체 누리호 4차 발사가 성공적으로 이뤄지며 우주 기술 자립의 기반을 다졌다.",
        metadata={"created_at": "2024-11-01"},
    ),
    Document(
        page_content="12월: 정부가 저출산 문제 해결을 위한 파격적인 지원 정책들을 발표하며 사회적 논의가 활발해졌다.",
        metadata={"created_at": "2024-12-01"},
    ),
]
```

```python
# FAISS 벡터 저장소 생성
vectorstore = FAISS.from_documents(docs, OpenAIEmbeddings())

# TimeWeightedVectorStoreRetriever 설정
# decay_rate: 시간 경과에 따른 중요도 감소율. 0에 가까울수록 최신 문서에 큰 가중치를 둡니다.
# retriever = TimeWeightedVectorStoreRetriever(
#     vectorstore=vectorstore, decay_rate=0.99, k=5
# )
retriever = TimeWeightedVectorStoreRetriever(
    vectorstore=vectorstore, decay_rate=0.1, k=5, other_score_keys=["final_score"]
)
```

```python
# '정부 정책' 관련 문서 검색
query = "정부 정책"
# invoke 메소드에 'now' 파라미터로 현재 시간을 전달합니다.
retrieved_docs = retriever.invoke(query, now=datetime.now())
```
