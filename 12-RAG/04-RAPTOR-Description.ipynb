# RAPTOR 
RAPTOR는 대규모 문서에서 복잡하고 광범위한 질문에 대한 답변 품질을 높이기 위해 설계된 고급 검색 기법입니다. 기존의 검색 증강 생성(RAG)이 문서의 작은 조각(chunk) 단위로만 정보를 찾는 한계를 극복하기 위해, 문서들을 재귀적으로 클러스터링하고 요약하여 계층적인 트리 구조로 정리하는 방식을 사용합니다.

## RAPTOR의 작동 방식
RAPTOR는 텍스트를 나무와 같은 구조로 만들어, 질문의 수준에 따라 나무의 잎(세부 정보)부터 줄기(핵심 요약), 뿌리(전체 주제)까지 탐색할 수 있게 합니다. 이 과정은 크게 3단계로 이루어집니다.

1. 클러스터링 (Clustering): 먼저 문서를 작은 텍스트 조각으로 나눈 뒤, 의미적으로 유사한 조각들을 그룹으로 묶습니다 (클러스터링). 예를 들어, 한 문서에서 'AI의 역사'에 대한 부분과 'AI의 최신 기술'에 대한 부분을 각각 다른 그룹으로 묶는 것과 같습니다.

2. 요약 (Summarization): 각 클러스터(그룹)의 내용을 요약하여 더 높은 수준의 추상적인 정보를 담은 새로운 텍스트 조각을 생성합니다. 이렇게 생성된 요약문은 여러 세부 정보들을 하나로 아우르는 핵심 개념이 됩니다.

3. 재귀적 구축 (Recursive Building): 1단계와 2단계를 반복합니다. 즉, 생성된 요약문들을 다시 클러스터링하고, 그 결과를 또 요약하는 과정을 더 이상 새로운 클러스터를 만들 수 없을 때까지 재귀적으로 수행합니다. 이 과정을 통해 세부 텍스트 조각(잎)부터 중간 요약(가지), 최종 요약(줄기)에 이르는 계층적 트리 구조가 완성됩니다.

## 정보 검색 및 활용
이렇게 만들어진 트리 구조를 활용해 사용자의 질문에 가장 적합한 정보를 찾습니다.

Tree Traversal: 질문과 관련된 정보를 찾기 위해 트리의 최상위 노드(가장 추상적인 요약)부터 시작하여 가장 관련성 높은 하위 노드로 이동하며 세부 정보를 찾아갑니다.

Collapsed Tree: 트리의 모든 노드(세부 정보, 중간 요약, 최종 요약)를 한곳에 모아놓고 질문과 가장 관련성이 높은 정보를 한 번에 검색합니다.

### UMAP 
고차원의 데이터를 저차원의 데이터로 변환하는 차원 축소(Dimensionality Reduction) 기법입니다.

이해하기 쉬운 비유는 지구본을 평면 세계 지도로 만드는 것입니다. 3차원 구체인 지구본의 정보를 2차원 평면에 옮기면서도, 각 대륙과 국가 간의 상대적인 위치 관계(구조)를 최대한 유지하려고 노력하죠.

마찬가지로 <u>UMAP은 텍스트 임베딩(수백 개의 숫자로 표현된 텍스트의 의미 벡터)처럼 매우 높은 차원의 데이터를 2차원이나 3차원처럼 훨씬 낮은 차원으로 압축합니다.</u> 이때 중요한 것은 원래 데이터가 가진 '의미적 구조'를 최대한 보존하는 것입니다. 즉, 원래 공간에서 의미가 비슷했던 텍스트들은 차원이 축소된 새로운 공간에서도 여전히 가까이 모여있게 됩니다.

## RAPTOR에서 UMAP의 작동 과정
RAPTOR의 트리 구축 과정에서 UMAP은 다음과 같은 역할을 합니다.

임베딩 생성: 문서의 모든 텍스트 조각들을 고차원의 벡터(임베딩)로 변환합니다.

UMAP 적용 (차원 축소): UMAP을 사용해 이 고차원 임베딩들을 클러스터링이 용이한 저차원(예: 10차원) 공간으로 압축합니다. 이때 원본 데이터의 의미적 유사성 구조가 보존됩니다.

클러스터링: 차원이 축소된 새로운 공간에서 HDBSCAN과 같은 클러스터링 알고리즘을 실행하여 비슷한 텍스트 조각들을 그룹으로 묶습니다. UMAP 덕분에 클러스터링의 성능과 정확도가 크게 향상됩니다.

요약 및 상위 노드 생성: 각 클러스터에 속한 텍스트 조각들을 요약하여 새로운 상위 레벨의 노드(부모 노드)를 만듭니다.

결론적으로, UMAP은 RAPTOR가 문서의 의미적 구조를 효과적으로 파악하고, 이를 바탕으로 정확한 클러스터를 형성하여 계층적 요약 트리를 성공적으로 구축할 수 있게 해주는 핵심적인 다리 역할을 합니다.
