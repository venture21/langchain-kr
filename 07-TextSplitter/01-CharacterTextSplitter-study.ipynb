### CharacterTextSplitter
1. CharacterTextSplitter 특성을 정의하고, CharacterTextSplitter객체를 생성
2. create_documents()함수로 문서와 연결+나누기(Documents로 변환해서 나누기)
3. split_text()함수를 사용해서 문서를 나눈다.(일반 텍스트를 나누기)

참고 문서 페이지
https://python.langchain.com/api_reference/text_splitters/character/langchain_text_splitters.character.CharacterTextSplitter.html

```python
# 0. txt파일 읽어오기
with open("./data/appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read()
```

```python
print(file[:500])
```

```python
# 1. CharacterTextSplitter 특성을 정의하고, CharacterTextSplitter객체를 생성
from langchain.text_splitter import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="\n\n",  # 문장을 나누는 구분자 설정
    chunk_size=250,  # 각 청크의 최대 길이 설정
    chunk_overlap=0,  # 청크 간의 겹치는 부분 설정
    length_function=len,  # 길이를 계산하는 함수 설정
)
```

```python
# 2. create_documents()함수로 문서와 연결+splitting
#    file은 일반 문서이므로, document로 변환
texts = text_splitter.create_documents([file])
```

```python
texts[0]
```

```python
print(texts[0].page_content)
```

```python
fieName = "./data/appendix-keywords-edit.txt"
# 파일 불러오기
with open(fieName, encoding="utf-8") as f:
    file1 = f.read()
texts1 = text_splitter.create_documents([file1])
print(texts1[0].page_content)
```

```python
texts1[0]
```

```python
# 3. split_text()함수를 사용해서 문서를 나눈다.
# txt파일을 읽었을때, create_documents()함수를 사용하지 않고,
# split_text()함수를 사용해서 문서를 나눌 수 있다.
myText = text_splitter.split_text(file1)
```
