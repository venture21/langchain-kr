```python
from dotenv import load_dotenv

load_dotenv()
```

```python
import os
import json
from serpapi import GoogleSearch

# SerpApi 클라이언트를 초기화하기 위한 파라미터를 설정합니다.
# 'engine'을 'google_shopping'으로 설정하는 것이 중요합니다.
params = {
    "engine": "google_shopping",
    "q": "아이폰 16 프로",  # 검색할 상품명을 입력합니다.
    "hl": "ko",  # 검색 언어를 한국어로 설정합니다.
    "gl": "kr",  # 검색 지역을 대한민국으로 설정합니다.
    "api_key": os.getenv("SERPAPI_API_KEY"),
}

# 파라미터를 사용하여 GoogleSearch 객체를 생성합니다.
search = GoogleSearch(params)

try:
    # 검색을 실행하고 결과를 딕셔너리 형태로 가져옵니다.
    results = search.get_dict()

    # 결과에 'shopping_results' 키가 있는지 확인하여 성공 여부를 판단합니다.
    if "shopping_results" in results:
        print(
            f"✅ 검색 성공! '{params['q']}'에 대한 쇼핑 정보를 출력합니다. (상위 3개)"
        )

        # 상위 3개 상품 정보만 추출합니다.
        top_3_items = results["shopping_results"][:3]

        # 각 상품의 주요 정보를 보기 쉽게 출력합니다.
        for i, item in enumerate(top_3_items, 1):
            print("\n" + "=" * 40)
            print(f"결과 {i}")
            print("=" * 40)
            print(f"상품명: {item.get('title')}")
            print(f"가격: {item.get('price')}")
            print(f"판매처: {item.get('source')}")
            print(f"리뷰 수: {item.get('reviews', '정보 없음')}")
            print(f"평점: {item.get('rating', '정보 없음')}")
            print(f"링크: {item.get('link')}")

        # 첫 번째 결과의 전체 JSON 데이터를 확인하고 싶을 경우 아래 코드의 주석을 해제하세요.
        # print("\n--- 첫 번째 상품 상세 정보 (JSON) ---")
        # print(json.dumps(top_3_items[0], indent=2, ensure_ascii=False))

    elif "error" in results:
        print(f"❌ 검색 중 오류가 발생했습니다: {results['error']}")
    else:
        print("⚠️ 쇼핑 정보를 찾을 수 없습니다. 다른 검색어로 시도해보세요.")
        print("\n전체 응답:")
        print(json.dumps(results, indent=2, ensure_ascii=False))

except Exception as e:
    print(f"스크립트 실행 중 예외가 발생했습니다: {e}")
```
